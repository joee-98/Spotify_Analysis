## Spotify Data Cleaning and Preparation using Python

### **Objective:** 
This notebook focuses on preparing the Spotify dataset for analysis. The goal is to clean the data by removing invalid records and fixing text inconsistencies so that it can be used for SQL-based analysis.

```python
# Import libraries

import pandas as pd
import numpy as np
```

### Loading the dataset to inspect its structure and understand the available columns.

```python
# Load the dataset

data=pd.read_csv('spotify_dataset.csv')
```

### Data Cleaning Steps

1. **Change datatypes**  
   Ensure each column has the correct format for accurate analysis.

2. **Remove unwanted columns**  
   Eliminate unnecessary data to simplify processing and improve performance.

3. **Remove null values**  
   Avoid missing data issues that can affect calculations and insights.

4. **Remove outliers**  
   Prevent extreme values from distorting analysis results.

5. **Remove duplicates**  
   Ensure each record is unique to maintain data accuracy.

6. **Remove whitespaces**  
   Clean text fields to avoid inconsistencies and matching errors.

7. **Modify rows for better processing**  
   Standardize data to make it easier for analysis and querying.

```python
# Change Boolean datatype to int

data['explicit']=data['explicit'].astype(int)
```

```python
# Remove unwanted columns

data.drop('Unnamed: 0',axis=1,inplace=True)
```

```python
# Check for null

data.isnull().sum()
```

```python
# Drop null values

data.dropna(axis=0,inplace=True)
```

```python
# Check for duplicates

data.duplicated().sum()
```

```python
# Remove duplicates

data.drop_duplicates(ignore_index=True, inplace=True)
```

```python
# Rename column for better readability

data.rename({'artists':'artist_name','key':'track_key','mode':'track_mode'},axis=1,inplace=True)
```

```python
# Text cleanup

data['track_name'] = data['track_name'].str.strip()
data['album_name'] = data['album_name'].str.strip()
data['artist_name'] = data['artist_name'].str.strip()
```

```python
# check for outliers
# danceability, energy, mode, speechiness,acousticness, explicit, instrumentalness, 
# liveness, valance all have values only between 0 and 1

data.describe()
```

```python
# Change duration from ms to mins

data['duration_mins']=(data['duration_ms']/60000).round(1)
```

```python
# Make all clumn_names small letter for better sql query execution

data.columns = data.columns.str.lower()
```

```python
# Clean column names by removing whitespaces

data.columns = data.columns.str.strip()
```

```python
# Ensure track_id is string before using .str

data['track_id'] = data['track_id'].astype(str)
```

```python
# Remove bad rows (very long corrupted ones)

data = data[data['track_id'].str.len() < 50]
```

```python
# Remove line breaks inside data

data = data.replace({r'\n': ' ', r'\r': ' '}, regex=True)
```

```python
# Drop completely empty rows

data = data.dropna(how='all')
```

```python
# Final Check

data.info()
data.isnull().sum()
data.duplicated().sum()
```

### Exporting the cleaned dataset for further analysis in SQL.

```python
# Save clean file
data.to_csv("spotify_final_clean.csv", index=False)
```

### **Conclusion:**
The dataset has been cleaned by removing invalid rows and fixing text inconsistencies. The final dataset is now structured and ready for SQL-based analysis.
